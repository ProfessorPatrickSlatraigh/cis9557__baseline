## Example 3: Pre-Attentive Attributes -- Position vs. Color Saturation

**Discussion points**

- Pre-attentive attributes are visual properties processed by the visual system in under 250 milliseconds, before directed attention is applied.
- In Version 1, the highest-scoring unit is immediately identifiable because it is the longest bar. No scanning is required.
- In Version 2, the highest-scoring unit is encoded in color saturation. The bars are unsorted, so the audience must scan each bar to find the darkest one.
- The difference in reading speed is a function of visual architecture, not analytical sophistication. Version 1 does not require a more capable audience.
- Design question: which attribute is the most important piece of information the audience must extract? Encode that attribute in the most salient pre-attentive channel the chart type permits.

```python
units=['Unit F','Unit C','Unit A','Unit E','Unit B','Unit D']
scores=[77,91,82,85,74,68]
highlight='Unit C'
sp=sorted(zip(scores,units),reverse=True)
ss=[s for s,u in sp]; us=[u for s,u in sp]
fig,axes=plt.subplots(1,2,figsize=(13,5))
axes[0].barh(us,ss,color='#2166ac',alpha=0.85)
axes[0].set_xlim(0,100)
axes[0].set_title('Version 1: Key Category Encoded in Position\n(Sorted, Uniform Color)',fontsize=11)
axes[0].set_xlabel('Customer Satisfaction Score')
cv2=['#1a5276' if u==highlight else '#aed6f1' for u in units]
axes[1].barh(units,scores,color=cv2,alpha=0.90)
axes[1].set_xlim(0,100)
axes[1].set_title('Version 2: Key Category Encoded in Color Saturation\n(Unsorted)',fontsize=11)
axes[1].set_xlabel('Customer Satisfaction Score')
plt.suptitle('Pre-Attentive Attributes: Position vs. Color Saturation',fontsize=13)
plt.tight_layout(); plt.show()
```

## Example 4: Redundant Encoding

**Discussion points**

- Redundant encoding uses two or more pre-attentive attributes to convey the same piece of information simultaneously. In Version 2, the highest value is marked by both greater length and a different color hue.
- Legitimate use: redundant encoding improves accessibility for audiences with color vision deficiency, who cannot reliably distinguish hue but can still read length.
- Cost: redundant encoding draws disproportionate attention to a single bar, which can suppress attention to other analytically important features.
- When every bar is equally important to the argument, uniform color is the correct choice.

```python
categories=['Q1','Q2','Q3','Q4','Q5','Q6']
values=[42,58,51,73,49,55]
fig,axes=plt.subplots(1,2,figsize=(12,5))
axes[0].bar(categories,values,color='steelblue',alpha=0.85)
axes[0].set_title('Version 1: Single Encoding (Length Only)',fontsize=11)
axes[0].set_ylabel('Units Sold (000s)'); axes[0].set_ylim(0,90)
cr=['#d73027' if v==max(values) else 'steelblue' for v in values]
axes[1].bar(categories,values,color=cr,alpha=0.85)
axes[1].set_title('Version 2: Redundant Encoding (Length + Color Hue)',fontsize=11)
axes[1].set_ylabel('Units Sold (000s)'); axes[1].set_ylim(0,90)
plt.suptitle('Pre-Attentive Attributes: Redundant Encoding',fontsize=13)
plt.tight_layout(); plt.show()
```

## Example 5: Encoding Channel Hierarchy -- Pie Chart vs. Horizontal Bar Chart

**Discussion points**

- The Cleveland and McGill hierarchy establishes that position along a common scale is the most accurate channel for quantitative judgment; angle and area are substantially less accurate.
- In the pie chart, the five values are 26%, 23%, 21%, 17%, and 13%. The ranking is genuinely ambiguous from visual inspection of slice angles and areas alone.
- In the horizontal bar chart, the ranking is immediately and accurately apparent because the audience is reading position along a common scale.
- The pie chart and bar chart use identical data. The difference in readability is entirely a function of the encoding channel.
- This is the primary analytical objection to pie charts: they encode quantities in channels that produce systematic inaccuracy in rank judgment. This is a perceptual fact, not a stylistic preference.

```python
labels=['Segment A','Segment B','Segment C','Segment D','Segment E']
vals=[26,23,21,17,13]
c5=['#2166ac','#4dac26','#d01c8b','#f1a340','#998ec3']
fig,axes=plt.subplots(1,2,figsize=(13,5))
axes[0].pie(vals,labels=labels,colors=c5,autopct='%1.0f%%',startangle=90,textprops={'fontsize':10})
axes[0].set_title('Pie Chart: Ranking by Angle and Area',fontsize=11)
yp=np.arange(len(labels))
axes[1].barh(yp,vals,color=c5,alpha=0.85)
axes[1].set_yticks(yp); axes[1].set_yticklabels(labels)
axes[1].set_xlim(0,35); axes[1].set_xlabel('Market Share (%)')
axes[1].set_title('Bar Chart: Ranking by Position Along Common Scale',fontsize=11)
plt.suptitle('Encoding Channel Hierarchy: Pie Chart vs. Horizontal Bar Chart',fontsize=13)
plt.tight_layout(); plt.show()
```

## Example 6: Why 3D Charts Fail

**Discussion points**

- The Cleveland and McGill hierarchy places volume at the bottom of the accuracy ranking. Depth adds a third dimension that cannot be read accurately from a two-dimensional display.
- Perspective projection introduces a systematic bias: bars toward the back appear shorter than bars of equal value toward the front. The distortion is a function of viewing angle, not data.
- There is no analytical task for which a 3D bar chart produces more accurate quantitative judgment than a 2D bar chart.
- 3D charts persist because they appear sophisticated and are a software default. Both reasons are rooted in convention, not analytical defensibility.
- The correct professional response when a client requests a 3D chart: produce the 2D version and explain why it communicates the data more accurately.

```python
cats=['Product A','Product B','Product C','Product D','Product E']
v6=[45,62,38,71,55]
fig=plt.figure(figsize=(13,5))
ax1=fig.add_subplot(121,projection='3d')
xp=np.arange(len(cats))
ax1.bar3d(xp,np.zeros(5),np.zeros(5),0.6,0.6,v6,color='steelblue',alpha=0.75)
ax1.set_xticks(xp+0.3); ax1.set_xticklabels(cats,fontsize=8)
ax1.set_title('3D Bar Chart: Volume Encoding\n(Low Accuracy, Perspective Distortion)',fontsize=10)
ax1.set_zlabel('Units Sold (000s)',fontsize=8); ax1.set_yticks([])
ax2=fig.add_subplot(122)
ax2.bar(cats,v6,color='steelblue',alpha=0.85)
ax2.set_title('2D Bar Chart: Length Encoding\n(High Accuracy, No Distortion)',fontsize=10)
ax2.set_ylabel('Units Sold (000s)'); ax2.set_ylim(0,85)
plt.suptitle('3D vs. 2D: Accuracy of Visual Encoding',fontsize=13)
plt.tight_layout(); plt.show()
```

## Notebook 2: Part II -- The Perceptual Interpretation of Color and Shape    


### Section 1: The Distinction Between Perception and Interpretation

The research on visual perception covered earlier in this module — particularly the accuracy hierarchy established by Cleveland and McGill — addresses a specific and bounded question: how reliably can a viewer extract a quantitative value from a given visual encoding? A viewer reading a bar chart compares lengths. A viewer reading a pie chart estimates angles and areas. The perceptual question is whether that extraction is accurate, fast, and consistent across viewers.

**Interpretation is a separate cognitive process.** Once the viewer has extracted a value — correctly or otherwise — they assign meaning to it. That assignment is governed not by the geometry of the chart but by the associations the viewer brings to the visual elements used to encode the data. A viewer who accurately reads that one bar is 23 percent taller than its neighbor may simultaneously interpret that bar as representing something negative, something dangerous, or something belonging to a particular brand or institution — depending entirely on the color used to fill it.

These two processes — perception and interpretation — can align, reinforce each other, or conflict. A visualization designed exclusively for perceptual accuracy, without attention to the interpretive associations of its encoding choices, is a visualization with an unexamined blind spot. The analyst has solved half the communication problem.

The practical consequence is significant. A chart can be technically correct — accurate data, appropriate chart type, clean layout — and still produce incorrect conclusions in its audience because the color or shape encoding triggers associations that contradict the analyst's intended message. Identifying and managing those associations is a professional responsibility, not an aesthetic preference.

### Section 2: Sources of Color and Shape Association

Treating all color and shape associations as a single undifferentiated category obscures the analytical decisions that follow from them. Associations arise from four distinguishable sources, and the appropriate design response differs depending on which source is in play.

---

#### 2.1 Biological and Near-Universal Tendencies

Some color associations are grounded in perceptual physiology and are relatively stable across human populations. Warm colors — red, orange, yellow — are experienced as activating, advancing, and attention-commanding. Cool colors — blue, green — are experienced as receding and calming. This is not primarily a cultural artifact. It reflects the spectral sensitivity of the human visual system and has been documented across populations with limited shared cultural history.

These tendencies are reliable enough to be treated as defaults in the absence of contrary information about the audience. They do not override cultural convention, but they provide a baseline.

---

#### 2.2 Cultural Convention

Cultural convention is the most consequential source of variation for analysts working with international or multicultural audiences. The canonical example is red.

In the traffic infrastructure of most Western nations, red signals danger, prohibition, or required stopping. This association extends to emergency services, hazard warnings, and military signaling — the British Army's red coat, for example, was a convention of visibility and institutional identity that became associated with authority and, in contested contexts, with threat. In Western financial and industrial contexts, red has become the near-universal encoding for loss, decline, or failure.

In Chinese commercial, ceremonial, and cultural contexts, red carries the primary association of prosperity, celebration, and good fortune. It is the color of traditional wedding garments, of envelopes containing monetary gifts, and of festival decoration. A financial dashboard that uses red to indicate declining performance will be read correctly by a Western audience trained in financial convention and may produce the opposite interpretation in an audience whose cultural frame associates red with positive outcomes.

White presents a comparable divergence. In many Western contexts, white encodes cleanliness, neutrality, or absence. In several East Asian and South Asian contexts, white is the traditional color of mourning.

---

#### 2.3 Industry and Institutional Convention

Within specific domains, color conventions have become standardized to the point of functioning as a shared language among practitioners. In Western financial reporting, green encodes profit, gain, or exceeding target; red encodes loss, decline, or missing target. This convention is so deeply embedded that reversing it — using red to highlight exceptional positive performance — would produce reading errors even among experienced financial professionals.

Healthcare similarly uses red for critical alerts and abnormal values, with color-coded severity scales (green to yellow to red) appearing in clinical dashboards, early warning systems, and laboratory result displays.

Cartography provides a near-universal example: blue encodes water. This convention is sufficiently stable across geographic and cultural traditions that departing from it requires explicit justification.

---

#### 2.4 Brand and Institutional Association

Sustained marketing exposure creates color associations that operate independently of cultural or industry convention. An audience that has been exposed to decades of brand communication learns to associate a specific hue with a specific entity. The association can become strong enough to influence interpretation of unrelated visualizations that happen to use the same color.

The case of Coca-Cola and Pepsi-Cola illustrates the point with particular clarity. Both brands have staked claims to specific hues — Coca-Cola to a specific red, Pepsi to a specific blue — that are protected as intellectual property and managed as strategic assets. Research on consumer perception has documented that brand-associated colors activate brand-related associations even when the color appears in contexts unrelated to the brand. An analyst presenting competitive market share data who uses the brand colors of the competitors in the visualization is making an interpretive choice, not merely a decorative one. The audience's associations with those colors will be activated.

The practical implication is that an analyst preparing materials for presentation to a mixed competitive audience — or to an audience with strong brand loyalties — may need to use neutral color encodings rather than the colors that would otherwise be the most intuitive choices.

### Section 3: Analytical Consequences — Three Cases

Understanding the sources of color association is prerequisite to recognizing the three design cases that follow from them.

**Case 1: Alignment.** The analyst's intended meaning matches the audience's pre-existing association. Red for a declining metric in a Western financial context is the standard example. The convention reinforces the message. The audience expends no cognitive effort resolving a conflict between what the chart says and what the color implies. This is the low-friction case.

**Case 2: Conflict.** The analyst's intended meaning contradicts the audience's association. A company whose primary brand color is red presents its own financial performance on a dashboard that uses red to encode declining results. The brand association — red as identity, familiarity, and positive recognition — conflicts with the financial convention — red as loss. The audience must resolve the conflict. Some will resolve it correctly; others will not. The ambiguity is an artifact of the design choice, not of the underlying data.

**Case 3: Audience Heterogeneity.** The visualization is viewed by audiences whose associations differ systematically. An international performance dashboard that uses red to indicate risk and green to indicate performance on target will be read correctly by audiences whose cultural and professional frame aligns with Western financial convention. Members of an audience for whom red carries strong positive associations must consciously override that association to read the chart as intended. The cognitive cost is borne by those audience members, not by the designer — and the designer may not be aware that the cost exists.

The three cases are not merely theoretical. They arise in every presentation to a non-homogeneous audience. The analyst's task is to determine which case applies and to make the design choice that minimizes the risk of misinterpretation.

The code examples that follow illustrate Cases 1 and 2 directly.

#### Code Example 1: The Same Data, Three Color Encodings

The six business units below show identical performance data. The values are fixed. Only the color encoding changes across the three panels.

- **Panel A** uses the Western financial convention: green for above target, red for below target.
- **Panel B** uses a neutral encoding: blue for above target, orange for below target. Neither color carries a strong prior financial association.
- **Panel C** uses a brand-conflicted encoding: the company's brand red is used for the positive bars, inverting the financial convention.

Observe how your interpretation of the data shifts across the three panels, even though the bars are identical in height.

```python
# ── Dataset ────────────────────────────────────────────────────────────────
units      = ['Northeast', 'Southeast', 'Midwest', 'Southwest', 'West Coast', 'Mountain']
target     = 100  # index baseline
actuals    = [112, 88, 95, 103, 76, 108]  # performance vs. 100 target
above      = [v >= target for v in actuals]

# ── Color palettes ─────────────────────────────────────────────────────────
palettes = [
    {
        'label':  'Panel A: Western Financial Convention',
        'pos':    '#2E8B57',   # green  -- gain / above target
        'neg':    '#C0392B',   # red    -- loss / below target
        'note':   'Red = below target   |   Green = above target\n'
                  'Convention: standard in Western financial reporting',
    },
    {
        'label':  'Panel B: Neutral Encoding',
        'pos':    '#2471A3',   # blue   -- no strong financial prior
        'neg':    '#D35400',   # orange -- no strong financial prior
        'note':   'Blue = above target   |   Orange = below target\n'
                  'Neither color carries a dominant financial convention',
    },
    {
        'label':  'Panel C: Brand-Conflicted Encoding',
        'pos':    '#C0392B',   # brand red used for positive outcome
        'neg':    '#2471A3',   # brand blue used for shortfall
        'note':   'Brand red = above target   |   Brand blue = below target\n'
                  'Red as positive conflicts with financial convention',
    },
]

x = np.arange(len(units))

fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharey=True)
fig.suptitle(
    'Identical Data, Three Color Encodings\n'
    'Business Unit Performance vs. Target (Index = 100)',
    fontsize=13, fontweight='bold', y=1.02
)

for ax, palette in zip(axes, palettes):
    colors = [palette['pos'] if a else palette['neg'] for a in above]
    bars   = ax.bar(x, actuals, color=colors, edgecolor='white', linewidth=0.8, width=0.6)

    # Target line
    ax.axhline(target, color='#333333', linewidth=1.2, linestyle='--', label='Target = 100')

    # Value labels
    for bar, val in zip(bars, actuals):
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 0.8,
            str(val),
            ha='center', va='bottom', fontsize=9, color='#333333'
        )

    ax.set_title(palette['label'], fontsize=10, fontweight='bold', pad=8)
    ax.set_xticks(x)
    ax.set_xticklabels(units, rotation=30, ha='right', fontsize=8)
    ax.set_ylim(60, 125)
    ax.set_ylabel('Performance Index' if ax == axes[0] else '', fontsize=9)
    ax.set_xlabel(palette['note'], fontsize=7.5, color='#555555', labelpad=10)
    ax.spines[['top', 'right']].set_visible(False)
    ax.tick_params(axis='y', labelsize=8)

plt.tight_layout()
plt.savefig('color_encoding_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
print('Figure saved.')
```

**Discussion.**  
The three panels are perceptually identical: the bar heights convey the same quantitative relationships in all three. What changes is the interpretive signal. Panel A produces an immediate, automatic reading of which units are performing well and which are not, because the color encoding aligns with a convention the audience already holds. Panel B is interpretively neutral — the audience must consult the legend before assigning valence. Panel C creates friction: the instinct produced by the financial convention (red = problem) conflicts with the label on the legend (red = above target). That friction is not resolved by the data; it is introduced by the design.

The appropriate panel depends on the audience. Panel A is correct for a Western financial audience. Panel B is the safer choice for a heterogeneous or international audience. Panel C is incorrect for any audience in which the financial convention is active — and its use would require explicit remediation in the chart legend and in any verbal presentation accompanying it.

#### Code Example 2: Shape Association in Scatter Plot Encoding

Shape carries associative meaning independently of color. Triangle markers pointing downward are associated, in many Western visual traditions, with warning, decline, or instability — their use in road signage (yield, warning), financial charts (price decline), and hazard iconography is extensive. Circle markers carry no dominant prior association in most business contexts; they are perceptually neutral.

The plot below shows the same two-group categorical distinction encoded first with neutral shapes (circles for both groups, differentiated only by color) and then with loaded shapes (triangles pointing down for the below-target group, circles for the above-target group).

The underlying data — scatter positions, group membership — are identical across the two panels.

```python
rng = np.random.default_rng(42)

# ── Synthetic dataset: 40 products, two groups ─────────────────────────────
n_above = 24
n_below = 16

above_x = rng.uniform(55, 95, n_above)   # customer satisfaction score
above_y = rng.uniform(105, 145, n_above) # revenue index vs. prior year

below_x = rng.uniform(30, 70, n_below)
below_y = rng.uniform(60, 99, n_below)

# ── Colors ─────────────────────────────────────────────────────────────────
col_above = '#2471A3'   # blue -- neutral
col_below = '#C0392B'   # red  -- note: same color in both panels

fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=True)
fig.suptitle(
    'Shape Association in Categorical Encoding\n'
    'Customer Satisfaction vs. Revenue Index — 40 Products',
    fontsize=12, fontweight='bold', y=1.02
)

# ── Panel A: Neutral shapes (circles only) ─────────────────────────────────
ax = axes[0]
ax.scatter(above_x, above_y, marker='o', color=col_above, s=70,
           edgecolors='white', linewidths=0.5, label='Above target', zorder=3)
ax.scatter(below_x, below_y, marker='o', color=col_below, s=70,
           edgecolors='white', linewidths=0.5, label='Below target', zorder=3)
ax.set_title('Panel A: Neutral Shape Encoding\n(circles only, groups by color)', fontsize=10)
ax.set_xlabel('Customer Satisfaction Score', fontsize=9)
ax.set_ylabel('Revenue Index vs. Prior Year', fontsize=9)
ax.legend(fontsize=8, framealpha=0.9)
ax.spines[['top', 'right']].set_visible(False)
ax.tick_params(labelsize=8)

# ── Panel B: Loaded shapes (triangle-down for below-target group) ──────────
ax = axes[1]
ax.scatter(above_x, above_y, marker='o', color=col_above, s=70,
           edgecolors='white', linewidths=0.5, label='Above target (circle)', zorder=3)
ax.scatter(below_x, below_y, marker='v', color=col_below, s=80,
           edgecolors='white', linewidths=0.5, label='Below target (triangle down)', zorder=3)
ax.set_title('Panel B: Loaded Shape Encoding\n(triangle-down for below-target group)', fontsize=10)
ax.set_xlabel('Customer Satisfaction Score', fontsize=9)
ax.legend(fontsize=8, framealpha=0.9)
ax.spines[['top', 'right']].set_visible(False)
ax.tick_params(labelsize=8)

plt.tight_layout()
plt.savefig('shape_encoding_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
print('Figure saved.')
```

**Discussion.**  
In Panel A, the two groups are distinguished only by color. A viewer must consult the legend to assign meaning to the distinction. In Panel B, the shape of the below-target markers carries a prior association — the downward triangle is used in Western road signage, financial charting, and hazard iconography to suggest decline, caution, or instability. A viewer who brings that association to the chart begins interpreting the below-target group as problematic before reading the legend.

Whether this is appropriate depends on whether the analyst intends to signal that the below-target group requires attention. If the purpose of the visualization is diagnostic — directing the viewer's focus toward a group that requires intervention — the loaded shape reinforces the message. If the purpose is merely to distinguish two groups without implying a value judgment about either, the loaded shape introduces an unintended interpretive bias.

The same logic applies to upward and downward arrows, checkmarks and crosses, and any other shape with an established iconographic association. Using them deliberately is a design choice. Using them without recognizing their associative weight is a design error.

###<font color=blue> Section 4: Practical Guidance — A Design Checklist</font>

The four questions below constitute a minimum audit for any visualization in which color or shape is used to encode categorical meaning. They are sequenced to reflect the order in which the relevant information is typically available during the design process.
  
---
  
**Question 1: Who is the audience, and what conventions do they bring?**  
  
Identify whether the audience is culturally homogeneous or heterogeneous, and whether they share a professional convention (financial, clinical, cartographic, or other). If the audience is heterogeneous, treat cultural convention as unreliable and prefer neutral encodings.  
  
---
  
**Question 2: Do any of the colors or shapes you have selected carry a prior association that could conflict with your intended message?**  
  
Audit each color and shape choice against the four sources identified in Section 2: biological tendency, cultural convention, industry convention, and brand association. A conflict in any of these sources requires either a design revision or an explicit remediation in the legend and any accompanying verbal explanation.  
  
---
  
**Question 3: Is the color-to-meaning mapping stated explicitly in the visualization?**  
  
Do not rely on the audience to infer the convention. A legend is not optional when color encodes categorical meaning. The legend entry should state the mapping in unambiguous language: not merely a color swatch with a label, but a swatch paired with a complete description of what the color represents.  
  
---

**Question 4: Is the encoding consistent within the document?**  
  
Internal consistency is more important than alignment with any external convention. If blue encodes the above-target group in Chart 1, blue must encode the above-target group in Chart 5. A viewer who has learned the internal convention from the first chart will apply it to subsequent charts without re-reading the legend. Inconsistency between charts in the same document produces misreadings that the viewer may not detect.

Supplementing color with shape, pattern, or explicit numerical labels is the correction.</i>
