```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.dpi']=100
plt.rcParams['font.family']='sans-serif'
```

Examples 12 and 13 use the same dataset: customer satisfaction scores across eight business units. The two examples apply different chart types to the same comparison task.

```python
bus_units=['Corporate Banking','Retail Banking','Wealth Management',
           'Insurance','Mortgage','Auto Finance','Small Business','Card Services']
scores_bu=[84.2,76.8,91.3,79.5,72.1,88.6,81.4,69.7]
sp=sorted(zip(scores_bu,bus_units),reverse=True)
ss=[s for s,u in sp]; us=[u for s,u in sp]
```

---
## Example 12: Horizontal Bar Chart for Categorical Comparison

**Discussion points**

- The horizontal bar chart is the most analytically reliable general-purpose chart for categorical comparison. Quantity is encoded in length along a common baseline.
- Horizontal orientation is preferred when category labels are long. Rotating vertical labels reduces readability.
- Sorting by value (descending) rather than alphabetically allows the audience to immediately identify rank order.
- The reference line at the company target (80) converts the chart from a description of scores into a performance evaluation.
- The y-axis must start at zero. A non-zero baseline distorts the apparent length differences among bars.

```python
fig,ax=plt.subplots(figsize=(9,6))
ax.barh(us,ss,color='steelblue',alpha=0.85)
ax.set_xlim(0,100)
ax.set_xlabel('Customer Satisfaction Score (0-100)')
ax.set_title('Customer Satisfaction by Business Unit: Q1 2026',fontsize=13)
ax.axvline(x=80,color='gray',linestyle='--',alpha=0.6,linewidth=1.5,
           label='Company target (80)')
ax.legend(fontsize=9)
plt.tight_layout(); plt.show()
```

## Example 13: Cleveland Dot Plot for Ranked Comparison

**Discussion points**

- The dot plot encodes the same information as the bar chart using a point on an axis rather than a bar from the baseline. Quantity is encoded in position along a common scale.
- Replacing the bar with a dot reduces ink and visual mass. With many categories, this reduction can make patterns easier to read.
- The horizontal guide line is a visual aid, not a data encoding. It guides the eye from label to value.
- Practical limitation: the dot plot is less familiar to general management audiences. An analyst using it must be prepared to briefly explain the reading convention.
- Both charts convey identical information. The choice is a function of audience familiarity and number of categories.

```python
fig,ax=plt.subplots(figsize=(9,6))
for i,(score,unit) in enumerate(zip(ss,us)):
    ax.plot([60,score],[i,i],color='lightgray',linewidth=1.5,zorder=1)
    ax.scatter(score,i,color='steelblue',s=80,zorder=2)
    ax.text(score+0.8,i,f'{score:.1f}',va='center',fontsize=9)
ax.set_xlim(60,100)
ax.set_ylim(-0.5,len(us)-0.5)
ax.set_yticks(range(len(us))); ax.set_yticklabels(us,fontsize=9)
ax.set_xlabel('Customer Satisfaction Score (0-100)')
ax.set_title('Customer Satisfaction by Business Unit: Dot Plot (Same Data as Example 12)',fontsize=11)
ax.axvline(x=80,color='gray',linestyle='--',alpha=0.6,linewidth=1.5)
ax.spines['top'].set_visible(False); ax.spines['right'].set_visible(False)
plt.tight_layout(); plt.show()
```

## Example 14: Slope Chart

**Discussion points**

- The slope chart encodes two values per entity connected by a line. The angle encodes direction and approximate magnitude of change.
- Blue lines indicate gained share; red lines indicate lost share. This redundant encoding makes winners and losers immediately apparent without requiring numerical comparison.
- The slope chart is superior to a grouped bar chart when the story is about which entities changed direction, not absolute values at each time point.
- Note the rank crossing between Gamma LLC and Beta Inc. That crossing is the analytically significant event in the chart, and it is visually foregrounded by the slope format.
- The slope chart is not appropriate for more than two time points. With three or more, a line chart or bump chart applies.

```python
comps=['Alpha Corp','Beta Inc','Gamma LLC','Delta Co','Epsilon Ltd','Zeta Group']
sy1=[28.4,22.1,18.3,14.2,10.5,6.5]
sy2=[31.2,19.8,20.1,12.8,9.7,6.4]
fig,ax=plt.subplots(figsize=(8,6))
for c,s1,s2 in zip(comps,sy1,sy2):
    col='#2166ac' if s2>=s1 else '#d6604d'
    ax.plot([0,1],[s1,s2],marker='o',linewidth=2,color=col,markersize=7)
    ax.text(-0.03,s1,f'{c}  {s1}%',ha='right',va='center',fontsize=9)
    ax.text(1.03,s2,f'{s2}%  {c}',ha='left',va='center',fontsize=9)
ax.set_xlim(-0.65,1.65); ax.set_ylim(4,35)
ax.set_xticks([0,1]); ax.set_xticklabels(['Year 1','Year 2'],fontsize=12)
ax.set_ylabel('Market Share (%)')
ax.set_title('Market Share by Competitor: Year 1 to Year 2\n(Blue = gained  |  Red = lost)',fontsize=12)
ax.spines['top'].set_visible(False); ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)
plt.tight_layout(); plt.show()
```

## Example 15: Small Multiples

**Discussion points**

- Small multiples display the same chart structure for each segment, arranged for side-by-side visual comparison. Each panel is identically scaled.
- Identical scaling across all panels is a non-negotiable requirement. If each panel auto-scales independently, cross-panel pattern comparison is invalid.
- With identical scaling, the Mountain region is immediately identifiable as the smallest in absolute revenue. The West Coast is the largest. These comparisons are available at a glance.
- Small multiples are appropriate when the analytical question involves comparing patterns across many segments. A single six-series chart would be overloaded.
- The cost of small multiples is page or screen real estate.

```python
regions=['Northeast','Southeast','Midwest','Southwest','West Coast','Mountain']
mab=list('JFMAMJJASOND')
rd={'Northeast':[42,38,45,48,52,55,54,56,58,62,65,70],
    'Southeast':[35,33,38,42,48,55,60,58,52,45,40,38],
    'Midwest':  [30,28,32,38,44,48,49,47,43,40,36,34],
    'Southwest':[45,43,46,50,54,58,61,60,57,53,49,47],
    'West Coast':[55,52,58,62,66,70,72,71,68,65,62,60],
    'Mountain': [22,20,24,28,33,38,40,39,35,30,25,23]}
fig,axes=plt.subplots(2,3,figsize=(13,7),sharey=True)
axes=axes.flatten()
for i,(region,ax) in enumerate(zip(regions,axes)):
    ax.bar(range(12),rd[region],color='steelblue',alpha=0.80)
    ax.set_title(region,fontsize=10,fontweight='bold')
    ax.set_xticks(range(12)); ax.set_xticklabels(mab,fontsize=7)
    ax.set_ylim(0,80)
    if i%3==0: ax.set_ylabel('Revenue ($ Millions)',fontsize=8)
plt.suptitle('Monthly Revenue by Region: Identical Scale Across All Panels',fontsize=12)
plt.tight_layout(); plt.show()
```

## Example 16: Bullet Chart vs. Gauge Chart

**Discussion points**

- The bullet chart encodes the actual value in length along a common scale, the most accurate encoding channel in the Cleveland and McGill hierarchy. The gauge encodes the same value in angle, a substantially less accurate channel.
- The bullet chart communicates three distinct pieces of information simultaneously for each business unit: the actual value, the target, and the qualitative performance band in which the actual value falls. The gauge communicates one value for one unit at a time without the qualitative band context.
- All 8 business units are visible and comparable in the bullet chart. Producing an equivalent gauge display for all 8 units would require 8 separate gauge panels, each of which can still only show one metric in a low-accuracy encoding channel.
- The target marker in the bullet chart is a white vertical line crossing the feature bar. The audience can immediately determine whether the actual value falls to the left (below target) or right (above target) of the marker. In the gauge, the target marker is a radial line on the arc; the judgment of whether the needle falls before or after that marker requires angle discrimination, which is less accurate.
- The gauge persists in business practice because it resembles a familiar physical instrument and appears dynamic. Neither of those properties is an analytical justification. The bullet chart was designed specifically to replace the gauge in performance reporting contexts, and it is analytically superior in every respect relevant to quantitative judgment.
- Discussion: in what presentation contexts might an analyst choose a gauge despite its analytical limitations, and what is the professional cost of that choice?

```python
# Example 16: Bullet Chart vs. Gauge Chart
# Left panel: bullet chart for all 8 business units -- Q1 sales actual vs. target.
# Right panel: gauge chart for one representative unit (Wealth Management).
# Same 8 business units as Examples 12 and 13; measure is quarterly sales, not satisfaction.
# Key analytical point: bullet chart encodes quantity in length (high accuracy);
# gauge encodes quantity in angle (lower accuracy). Bullet also shows qualitative
# performance bands and multi-unit comparison simultaneously -- gauge cannot.

import matplotlib.patches as mpatches
import matplotlib.lines as mlines
import numpy as np

units = [
    'Corporate Banking', 'Retail Banking', 'Wealth Management',
    'Insurance', 'Mortgage', 'Auto Finance', 'Small Business', 'Card Services'
]

# Q1 actual sales and targets ($ millions); values span all three qualitative zones
actuals = [38, 41, 59, 25, 47, 28, 22, 34]
targets = [42, 38, 55, 29, 47, 33, 25, 31]

MAX_VAL    = 70      # fixed x-axis scale applied identically to all units
BELOW_FRAC = 0.75    # below this fraction of target = below plan
APPR_FRAC  = 0.95    # below this fraction of target = approaching plan

NAVY   = '#0A1E35'
TEAL   = '#1C7293'
RED_M  = '#d6604d'   # target marker on gauge
C_POOR  = '#B0B0B0'  # darkest band: below plan
C_SATIS = '#CFCFCF'  # middle band: approaching plan
C_GOOD  = '#E8E8E8'  # lightest band: on / above plan

fig = plt.figure(figsize=(15, 6))
gs  = fig.add_gridspec(1, 2, width_ratios=[2.8, 1.2], wspace=0.30)
ax_b = fig.add_subplot(gs[0])
ax_g = fig.add_subplot(gs[1])

# ---- Left panel: Bullet chart -----------------------------------------------
ROW_H  = 0.60   # height of background performance bands
FEAT_H = 0.30   # height of actual-value feature bar (narrower than bands)

for i, (unit, actual, target) in enumerate(zip(units, actuals, targets)):
    below_t = BELOW_FRAC * target
    appr_t  = APPR_FRAC  * target

    # Paint background bands from full width inward (lightest zone last painted on top)
    ax_b.barh(i, MAX_VAL, height=ROW_H, color=C_GOOD,  align='center', zorder=1)
    ax_b.barh(i, appr_t,  height=ROW_H, color=C_SATIS, align='center', zorder=2)
    ax_b.barh(i, below_t, height=ROW_H, color=C_POOR,  align='center', zorder=3)

    # Feature bar: actual value (narrow, navy)
    ax_b.barh(i, actual, height=FEAT_H, color=NAVY, align='center', zorder=4)

    # Target marker: white vertical line crossing the feature bar
    ax_b.plot(
        [target, target],
        [i - ROW_H / 2 + 0.04, i + ROW_H / 2 - 0.04],
        color='white', linewidth=2.5, solid_capstyle='butt', zorder=5
    )

    # Actual value label to the right of the feature bar
    ax_b.text(actual + 0.9, i, f'${actual}M', va='center', ha='left',
              fontsize=8, color=NAVY, fontweight='bold')

ax_b.set_yticks(range(len(units)))
ax_b.set_yticklabels(units, fontsize=9)
ax_b.set_xlim(0, MAX_VAL)
ax_b.set_ylim(-0.7, len(units) - 0.3)
ax_b.set_xlabel('Q1 Sales ($ Millions)', fontsize=9)
ax_b.set_title(
    'Bullet Chart: All 8 Business Units\n'
    'Actual vs. Target with Qualitative Performance Bands',
    fontsize=10
)
ax_b.spines['top'].set_visible(False)
ax_b.spines['right'].set_visible(False)
ax_b.grid(axis='x', alpha=0.2, linewidth=0.5)

legend_handles = [
    mpatches.Patch(color=C_POOR,  label='Below plan  (< 75% of target)'),
    mpatches.Patch(color=C_SATIS, label='Approaching plan  (75-95% of target)'),
    mpatches.Patch(color=C_GOOD,  label='On / above plan  (>= 95% of target)'),
    mpatches.Patch(color=NAVY,    label='Actual value'),
    mlines.Line2D([], [], color='#888888', linewidth=2.5,
                  label='Target  (white marker on bars)'),
]
ax_b.legend(handles=legend_handles, loc='lower right', fontsize=7.5, framealpha=0.9)

# ---- Right panel: Gauge chart (Wealth Management only) ----------------------
# Wealth Management: actual=59, target=55 -- above plan; chosen as representative unit.
G_UNIT   = 'Wealth Management'
G_ACTUAL = 59
G_TARGET = 55
G_BELOW  = BELOW_FRAC * G_TARGET   # 41.25
G_APPR   = APPR_FRAC  * G_TARGET   # 52.25

def to_angle(v):
    # Maps 0 -> 180 degrees (left / minimum), MAX_VAL -> 0 degrees (right / maximum)
    return 180.0 * (1.0 - v / MAX_VAL)

ang_below  = to_angle(G_BELOW)    # ~74.0
ang_appr   = to_angle(G_APPR)     # ~45.6
ang_target = to_angle(G_TARGET)   # ~38.6
ang_actual = to_angle(G_ACTUAL)   # ~28.3

ax_g.set_xlim(-1.2, 1.2)
ax_g.set_ylim(-0.42, 1.18)
ax_g.set_aspect('equal')
ax_g.axis('off')

R_OUT = 1.00
R_IN  = 0.58

def draw_wedge(ax, t1, t2, color):
    w = mpatches.Wedge(
        (0, 0), R_OUT, t1, t2, width=R_OUT - R_IN, color=color, zorder=2
    )
    ax.add_patch(w)

# Zone arcs (drawn counterclockwise from t1 to t2)
draw_wedge(ax_g, ang_below, 180,       C_POOR)   # below plan
draw_wedge(ax_g, ang_appr,  ang_below, C_SATIS)  # approaching plan
draw_wedge(ax_g, 0,         ang_appr,  C_GOOD)   # on / above plan

# Target marker: radial line on the arc in red-orange
tx = np.cos(np.radians(ang_target))
ty = np.sin(np.radians(ang_target))
ax_g.plot(
    [R_IN * tx, R_OUT * tx],
    [R_IN * ty, R_OUT * ty],
    color=RED_M, linewidth=2.8, solid_capstyle='butt', zorder=5
)

# Needle pointing to actual value
nx = np.cos(np.radians(ang_actual))
ny = np.sin(np.radians(ang_actual))
ax_g.annotate(
    '', xy=(0.82 * nx, 0.82 * ny), xytext=(0.05 * nx, 0.05 * ny),
    arrowprops=dict(arrowstyle='->', color=NAVY, lw=2.0, mutation_scale=12)
)

# Center dot
ax_g.add_patch(mpatches.Circle((0, 0), 0.07, color=NAVY, zorder=6))

# Scale labels on outer arc edge
for val, lbl in [(0, '0'), (35, '35'), (MAX_VAL, str(MAX_VAL))]:
    ang = to_angle(val)
    lx = 1.14 * np.cos(np.radians(ang))
    ly = 1.14 * np.sin(np.radians(ang))
    ax_g.text(lx, ly, lbl, ha='center', va='center', fontsize=7.5, color='#555555')

# Value callouts below center
ax_g.text(0, -0.17, f'Actual: ${G_ACTUAL}M', ha='center',
          fontsize=9, color=NAVY, fontweight='bold')
ax_g.text(0, -0.28, f'Target: ${G_TARGET}M', ha='center',
          fontsize=8.5, color=RED_M)
ax_g.text(0, -0.40,
          'Shows 1 of 8 units only.\nBands are less precisely readable.',
          ha='center', fontsize=7.5, color='#666666', style='italic')

ax_g.set_title(
    f'Gauge Chart: {G_UNIT} Only\n(Angle encoding; one unit at a time)',
    fontsize=10
)

plt.suptitle(
    'Example 16: Bullet Chart vs. Gauge Chart -- Same Data, Different Encoding Channels',
    fontsize=12, y=1.01
)
plt.tight_layout()
plt.show()
```
