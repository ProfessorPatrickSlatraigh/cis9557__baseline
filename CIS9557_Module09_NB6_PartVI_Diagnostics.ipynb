## Example 22: Truncated Y-Axis

**Discussion points**

- In Version 1, the six regional scores cluster between 77.8 and 82.4 -- a spread of 4.6 points on a 100-point scale. The chart accurately conveys that the differences are modest.
- In Version 2, the y-axis begins at 75. Mountain appears roughly half the height of Northeast even though the actual score difference is less than 5 points.
- This is not a scale choice; it is a distortion. The audience takes bar height as proportional to the underlying quantity. Truncating the axis breaks that proportionality.
- A truncated y-axis is acceptable for line charts and dot plots. It is not acceptable for bar charts.
- When reviewing AI-generated or software-generated bar charts, the first diagnostic check is whether the y-axis starts at zero.

```python
regs=['Northeast','Southeast','Midwest','Southwest','West Coast','Mountain']
sc=[82.4,79.1,80.6,78.3,81.9,77.8]
fig,axes=plt.subplots(1,2,figsize=(13,5))
axes[0].bar(regs,sc,color='steelblue',alpha=0.85)
axes[0].set_ylim(0,100)
axes[0].set_title('Version 1: Y-Axis Begins at Zero',fontsize=11)
axes[0].set_ylabel('Customer Satisfaction Score')
axes[0].tick_params(axis='x',rotation=30)
axes[1].bar(regs,sc,color='steelblue',alpha=0.85)
axes[1].set_ylim(75,85)
axes[1].set_title('Version 2: Y-Axis Begins at 75 (Truncated)',fontsize=11)
axes[1].set_ylabel('Customer Satisfaction Score')
axes[1].tick_params(axis='x',rotation=30)
plt.suptitle('Example 22: Truncated Y-Axis -- Same Data, Two Different Impressions',fontsize=13)
plt.tight_layout(); plt.show()
```

## Example 23: Line Chart Applied to Categorical Data

**Discussion points**

- The six product categories have no natural ordering and no continuous relationship. There is no reason to believe the return rate changes continuously between Laptops and Tablets.
- The connecting line in Version 1 implies that a continuous relationship exists. This implication is introduced by the chart type, not the data.
- Audiences trained in line chart reading will tend to interpret the slope as directionally meaningful. In this case, those interpretations are unsupported by the data structure.
- The test: if reordering the categories would produce a different-looking chart without changing the data, the connecting line is unjustified. Categorical data can always be reordered; time series data cannot.

```python
prod_cats=['Laptops','Tablets','Monitors','Keyboards','Mice','Headsets']
ret=[4.2,3.8,2.1,1.4,0.9,2.7]
fig,axes=plt.subplots(1,2,figsize=(13,5))
axes[0].plot(prod_cats,ret,marker='o',linewidth=2,color='steelblue')
axes[0].set_title('Version 1: Line Chart on Categorical Data (Incorrect)\n'
                  'Connecting line implies continuity that does not exist',fontsize=10)
axes[0].set_ylabel('Return Rate (%)')
axes[0].set_ylim(0,6); axes[0].tick_params(axis='x',rotation=25)
axes[1].bar(prod_cats,ret,color='steelblue',alpha=0.85)
axes[1].set_title('Version 2: Bar Chart for Categorical Data (Correct)',fontsize=10)
axes[1].set_ylabel('Return Rate (%)')
axes[1].set_ylim(0,6); axes[1].tick_params(axis='x',rotation=25)
plt.suptitle('Example 23: Line Chart vs. Bar Chart for Categorical Data',fontsize=13)
plt.tight_layout(); plt.show()
```

## Example 24: Overloaded Multi-Series Chart

**Discussion points**

- Eight series with eight colors require the audience to decode each line by color before any data interpretation begins. Color-matching precedes analysis.
- Color hue is a low-accuracy encoding channel for distinguishing among more than seven categories. Some pairs will be nearly indistinguishable under standard lighting or in print.
- Alternatives: (1) small multiples on identical scales; (2) highlight one series against muted background lines; (3) aggregate into fewer meaningful groups.
- The chart is technically correct. It is analytically inadequate because no finding can be extracted without extended effort.
- Discussion: what would the analyst need to know about the business question before deciding which alternative to use?

```python
np.random.seed(42)
m24=[f'M{i+1:02d}' for i in range(24)]
pls=['Alpha','Beta','Gamma','Delta','Epsilon','Zeta','Eta','Theta']
c8=['#e41a1c','#377eb8','#4daf4a','#984ea3','#ff7f00','#a65628','#f781bf','#999999']
fig,ax=plt.subplots(figsize=(12,6))
for i,pl in enumerate(pls):
    base=20+i*5
    vals=base+np.cumsum(np.random.normal(0,2.5,24))
    ax.plot(range(24),vals,label=pl,color=c8[i],linewidth=1.8)
ax.set_xticks(range(0,24,4))
ax.set_xticklabels([m24[i] for i in range(0,24,4)])
ax.set_title('Monthly Revenue by Product Line: 8 Series, 24 Months',fontsize=13)
ax.set_ylabel('Revenue ($ Millions)')
ax.legend(loc='upper left',fontsize=8,ncol=2)
plt.tight_layout(); plt.show()
```

## Example 25: Descriptive Title vs. Analytical Title

**Discussion points**

- 'Customer Satisfaction by Region, Q1 2026' describes the chart. It tells the audience what the data covers. It does not tell the audience what the data means.
- 'Southeast Trails All Regions by 10+ Points for Third Consecutive Quarter' communicates the finding. The conclusion is stated before the audience examines the chart in detail.
- The analytical title converts the chart from a display into an argument. A chart that makes an explicit argument is more useful to a decision-making audience.
- The 'weather report' problem: a descriptive title is the visual equivalent of a weather report. Executives need the conclusion and its implication for a decision, not a description of the data.
- Producing an analytical title requires the analyst to commit to a specific interpretation. An analyst who cannot state the finding in a title has not finished the analysis.

```python
regs25=['Northeast','Southeast','Midwest','Southwest','West Coast','Mountain']
sc25=[88.2,74.1,82.5,79.8,85.3,80.1]
col25=['#2166ac' if s>=80 else '#d6604d' for s in sc25]
fig,axes=plt.subplots(1,2,figsize=(14,5))
for ax in axes:
    ax.barh(regs25,sc25,color=col25,alpha=0.85)
    ax.set_xlim(0,100); ax.set_xlabel('Customer Satisfaction Score')
    ax.axvline(x=80,color='gray',linestyle='--',alpha=0.65,linewidth=1.5,
               label='Company target (80)')
    ax.legend(fontsize=9)
axes[0].set_title('Customer Satisfaction by Region, Q1 2026\n'
                  '(Descriptive: states what the data is)',fontsize=10)
axes[1].set_title('Southeast Trails All Regions by 10+ Points for Third Consecutive Quarter\n'
                  '(Analytical: states what the data means)',fontsize=10)
plt.suptitle('Example 25: The Same Chart, Two Different Titles',fontsize=13)
plt.tight_layout(); plt.show()
```
