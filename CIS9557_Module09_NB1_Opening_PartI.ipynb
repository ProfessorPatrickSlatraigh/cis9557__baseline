## Opening: Example 1 -- The Misread Chart

**Discussion points**

- The dataset has 12 months of net revenue. August shows a spike from a one-time contract settlement, not organic growth.
- The analyst intended to communicate the upward trend across the full year.
- In a meeting, audience attention goes to the August spike first. The trend -- the actual analytical finding -- becomes secondary.
- The visualization failed not because of bad data, but because the encoding gave more visual weight to an anomaly than to the intended argument.
- A chart is an argument. An uncontrolled chart is an uncontrolled argument.
- Discussion: what design choices would reduce the visual dominance of the spike while preserving an accurate factual representation of it?

```python
months=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
x=np.arange(12)
revenue=[4.2,4.5,4.3,4.6,4.8,4.7,4.9,7.1,5.0,5.2,5.4,5.6]
fig,ax=plt.subplots(figsize=(10,5))
ax.plot(x,revenue,marker='o',linewidth=2,color='steelblue',markersize=7)
ax.plot(7,7.1,marker='o',markersize=13,color='crimson',zorder=5)
ax.annotate('One-time contract settlement (August)',xy=(7,7.1),xytext=(9,6.6),
    arrowprops=dict(arrowstyle='->',color='crimson',lw=1.5),color='crimson',fontsize=10)
ax.set_xticks(x)
ax.set_xticklabels(months)
ax.set_title('Monthly Net Revenue: Fiscal Year ($ Millions)',fontsize=13)
ax.set_ylabel('Revenue ($ Millions)')
ax.set_ylim(0,8.5)
ax.axhline(y=4.2,color='gray',linestyle='--',alpha=0.4,linewidth=1,label='January baseline')
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()
```

## Part I: Example 2 -- Same Data, Three Different Questions

**Discussion points**

- All three charts use the identical dataset: quarterly revenue by product line over two fiscal years.
- Chart 2a answers: how is total revenue trending? The line chart encodes trajectory across time. It cannot show product line differences.
- Chart 2b answers: which product line performed best each quarter? The grouped bar encodes individual values per period. It cannot easily show aggregate totals.
- Chart 2c answers: how has revenue composition shifted? The 100% stacked bar encodes proportional structure. It removes the absolute total entirely.
- The data did not determine the chart. The analytical question did.
- An analyst who selects a chart before stating the question is making an analytical error, not a formatting choice.
- Each chart presents a valid truth about the same data. The professional responsibility is to select the one that matches the decision the audience must make.

```python
quarters=['Q1 Y1','Q2 Y1','Q3 Y1','Q4 Y1','Q1 Y2','Q2 Y2','Q3 Y2','Q4 Y2']
product_lines=['Software','Hardware','Services','Support']
data=np.array([[12.1,13.4,14.2,15.1,15.8,17.2,18.1,19.4],
               [ 8.3, 7.9, 8.1, 9.2, 7.8, 7.5, 7.2, 6.9],
               [ 5.2, 5.8, 6.1, 6.4, 6.9, 7.3, 7.8, 8.2],
               [ 3.1, 3.2, 3.0, 3.3, 3.2, 3.4, 3.5, 3.6]])
total_revenue=data.sum(axis=0)
palette4=['#2166ac','#d6604d','#1a9641','#e08a00']
```

```python
x8=np.arange(len(quarters))
fig,ax=plt.subplots(figsize=(9,5))
ax.plot(x8,total_revenue,marker='o',linewidth=2.5,color='steelblue',markersize=8)
ax.set_xticks(x8); ax.set_xticklabels(quarters)
ax.set_title('How Is Total Revenue Trending Over Time?',fontsize=13)
ax.set_ylabel('Total Revenue ($ Millions)')
ax.set_ylim(0,45); ax.grid(axis='y',alpha=0.3)
plt.tight_layout(); plt.show()
```

```python
x8=np.arange(len(quarters)); width=0.20
fig,ax=plt.subplots(figsize=(12,5))
for i,(pl,row) in enumerate(zip(product_lines,data)):
    ax.bar(x8+i*width,row,width,label=pl,color=palette4[i],alpha=0.85)
ax.set_xticks(x8+width*1.5); ax.set_xticklabels(quarters)
ax.set_title('Which Product Line Performed Best Each Quarter?',fontsize=13)
ax.set_ylabel('Revenue ($ Millions)')
ax.legend(fontsize=9); ax.set_ylim(0,25); ax.grid(axis='y',alpha=0.3)
plt.tight_layout(); plt.show()
```

```python
proportions=data/data.sum(axis=0)
bottom=np.zeros(len(quarters))
fig,ax=plt.subplots(figsize=(10,5))
for i,(pl,row) in enumerate(zip(product_lines,proportions)):
    ax.bar(quarters,row,bottom=bottom,label=pl,color=palette4[i],alpha=0.85)
    bottom+=row
ax.set_title('How Has Revenue Composition Shifted Across the Two Years?',fontsize=13)
ax.set_ylabel('Share of Total Revenue')
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y,_: f'{y:.0%}'))
ax.legend(fontsize=9,loc='upper right')
plt.tight_layout(); plt.show()
```
