## Scenario A: Revenue Share Across Time

A regional sales manager needs to present to the board how each of four product lines has changed in revenue share over the past two fiscal years. The company has been actively shifting its portfolio away from On-Premise Software toward Cloud Services. The board needs to understand the direction and rate of that shift, and how the other two lines have moved in the same period.

Apply the decision framework before writing any code.

```python
# Scenario A: Revenue Share Across Time
quarters_a=['Q1 Y1','Q2 Y1','Q3 Y1','Q4 Y1',
            'Q1 Y2','Q2 Y2','Q3 Y2','Q4 Y2']
product_lines_a=['Cloud Services','On-Premise Software',
                 'Professional Services','Support']
# Revenue in $ millions; rows = product lines, columns = quarters
revenue_a=np.array([
    [15.2,18.4,21.3,24.8,28.1,32.4,36.7,41.2],
    [32.4,30.1,27.8,25.5,23.2,21.0,18.8,16.7],
    [ 8.3, 9.1, 9.8,10.4,11.2,11.9,12.5,13.2],
    [ 5.6, 5.8, 5.7, 5.9, 6.0, 6.1, 6.2, 6.3],
])
```

```python
# SCENARIO A -- Student Work Cell
#
# Analytical question: How has the revenue share of each product line
# changed over the eight quarters?
#
# Step 1: State the decision dimension (Time, Comparison, or Composition).
# Step 2: Identify the specific chart type from the session framework.
# Step 3: State the primary pre-attentive attribute in which the key
#         variable will be encoded.
# Step 4: Produce the chart using the data provided above.
#
# Write your code below.
```

---
## Scenario A -- Instructor Solution

The cell below is the instructor solution. It should not be run or distributed to students prior to the class debrief.

```python
# ===========================================================
# SCENARIO A -- INSTRUCTOR SOLUTION
# ===========================================================
# Decision dimension: Composition (part-to-whole relationship across time).
# Chart type: BOTH are shown below with an explanation of when each applies.
#   100% Stacked Bar -- use when proportional composition is the question
#                        and absolute totals are not relevant to the audience.
#   Stacked Area     -- use when BOTH the aggregate growth trend AND the
#                        product mix shift are the analytical question.

from matplotlib.ticker import PercentFormatter

palette_a = ['#2166ac', '#d6604d', '#1a9641', '#e08a00']

fig, (ax_pct, ax_area) = plt.subplots(1, 2, figsize=(14, 5))

# ---- 100% Stacked Bar ---------------------------------------------------
proportions = revenue_a / revenue_a.sum(axis=0)
bottom = np.zeros(len(quarters_a))
for i, (pl, row) in enumerate(zip(product_lines_a, proportions)):
    ax_pct.bar(quarters_a, row, bottom=bottom, label=pl,
               color=palette_a[i], alpha=0.85)
    bottom += row
ax_pct.set_title(
    '100% Stacked Bar: Proportional Composition\n'
    'Use when: product mix shift is the question;\n'
    'absolute revenue totals are not relevant',
    fontsize=9
)
ax_pct.set_ylabel('Share of Total Revenue')
ax_pct.yaxis.set_major_formatter(PercentFormatter(xmax=1.0, decimals=0))
ax_pct.legend(fontsize=8, loc='upper right')
ax_pct.tick_params(axis='x', rotation=30)

# ---- Stacked Area -------------------------------------------------------
x8 = np.arange(len(quarters_a))
ax_area.stackplot(x8, revenue_a, labels=product_lines_a,
                  colors=palette_a, alpha=0.80)
ax_area.set_xticks(x8)
ax_area.set_xticklabels(quarters_a, rotation=30, ha='right', fontsize=8)
ax_area.set_title(
    'Stacked Area: Composition and Total Trajectory\n'
    'Use when: both aggregate growth and product mix\n'
    'shift are part of the analytical question',
    fontsize=9
)
ax_area.set_ylabel('Revenue ($ Millions)')
ax_area.set_ylim(0, 90)
ax_area.legend(fontsize=8, loc='upper left')

plt.suptitle(
    'Scenario A Solution: Revenue Share Across Time\n'
    'Chart type depends on whether absolute totals matter to the audience',
    fontsize=11, y=1.02
)
plt.tight_layout()
plt.show()
```

## Scenario B: Turnover Rate vs. Target by Department

An HR analyst is preparing a quarterly workforce review for the Chief People Officer. The report must show how each department's voluntary turnover rate compares to the company target and to qualitative performance thresholds.

Note that voluntary turnover rate is a lower-is-better metric. Apply the decision framework before writing any code, and consider how the directional convention of this metric affects the chart design.

```python
# Scenario B Data
# An HR analyst is preparing a quarterly workforce review for the Chief People Officer.
# The report must show how each department's voluntary turnover rate compares to the
# company target and to qualitative performance thresholds.
#
# IMPORTANT: voluntary turnover rate is a lower-is-better metric.
# The qualitative bands are arranged so that the leftmost zone (lowest turnover)
# represents the best performance. This is the inverse of the standard
# sales or revenue bullet chart convention.

departments = [
    'Product Development', 'Engineering', 'Sales', 'Marketing',
    'Finance', 'Human Resources', 'Operations', 'Customer Support'
]

# Voluntary turnover rate (%): actual and company-wide target
turnover_actual = [8.2, 11.5, 18.4, 14.2, 7.6, 9.8, 13.1, 16.7]
turnover_target = 12.0   # company target applied uniformly to all departments

# Qualitative thresholds (% voluntary turnover):
#   Below 8%   = excellent retention
#   8% to 12%  = acceptable (at or near target)
#   Above 12%  = concerning (above target, action required)
EXCELLENT_T = 8.0
MAX_RATE    = 22.0   # upper bound for the chart scale
```

```python
# SCENARIO B -- Student Work Cell
#
# Analytical question: How does each department's voluntary turnover rate
# compare to the company target and to qualitative performance thresholds?
#
# Note: turnover rate is a lower-is-better metric. The qualitative performance
# bands must be arranged so that the leftmost zone (lowest turnover)
# represents the best performance. This inverts the color convention used
# in Example 16, where higher values were better.
#
# Step 1: State the decision dimension (Time, Comparison, or Composition).
# Step 2: Identify the chart type from the session framework and explain
#         why it is analytically preferable to a simple bar chart with a
#         reference line at the target.
# Step 3: Describe how you will adapt the qualitative band encoding for a
#         lower-is-better metric and which encoding channel carries the
#         actual value, the target, and the qualitative context.
# Step 4: Produce the chart using the data provided above.
#
# Write your code below.
```

---
## Scenario B -- Instructor Solution

The cell below is the instructor solution. It should not be run or distributed to students prior to the class debrief.

```python
# ===========================================================
# SCENARIO B -- INSTRUCTOR SOLUTION
# ===========================================================
# Decision dimension: Comparison (actual vs. target vs. qualitative bands).
# Chart type: Bullet chart.
# Direction note: lower turnover is better. The qualitative band colors
# run from lightest (leftmost, lowest turnover, best) to darkest
# (rightmost, highest turnover, worst). This is the inverse of the
# standard sales-oriented bullet chart from Example 16.

import matplotlib.patches as mpatches
import matplotlib.lines as mlines

NAVY    = '#0A1E35'
C_GOOD  = '#E8E8E8'   # leftmost: excellent retention (< 8%)
C_SATIS = '#CFCFCF'   # middle: acceptable (8-12%)
C_POOR  = '#B0B0B0'   # rightmost: concerning (> 12%)

ROW_H  = 0.60
FEAT_H = 0.30

fig, ax = plt.subplots(figsize=(11, 6))

for i, (dept, actual) in enumerate(zip(departments, turnover_actual)):
    # Paint bands left to right: lightest (best) to darkest (worst)
    ax.barh(i, MAX_RATE,       height=ROW_H, color=C_POOR,  align='center', zorder=1)
    ax.barh(i, turnover_target, height=ROW_H, color=C_SATIS, align='center', zorder=2)
    ax.barh(i, EXCELLENT_T,    height=ROW_H, color=C_GOOD,  align='center', zorder=3)

    # Feature bar: actual turnover rate
    ax.barh(i, actual, height=FEAT_H, color=NAVY, align='center', zorder=4)

    # Target marker: white vertical line at company target
    ax.plot(
        [turnover_target, turnover_target],
        [i - ROW_H / 2 + 0.04, i + ROW_H / 2 - 0.04],
        color='white', linewidth=2.5, solid_capstyle='butt', zorder=5
    )
    ax.text(actual + 0.4, i, f'{actual}%', va='center', ha='left',
            fontsize=8, color=NAVY, fontweight='bold')

ax.set_yticks(range(len(departments)))
ax.set_yticklabels(departments, fontsize=9)
ax.set_xlim(0, MAX_RATE)
ax.set_ylim(-0.7, len(departments) - 0.3)
ax.set_xlabel('Voluntary Turnover Rate (%)', fontsize=9)
ax.set_title(
    'Scenario B Solution: Voluntary Turnover Rate by Department\n'
    'Lower is better -- band colors inverted relative to Example 16',
    fontsize=10
)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='x', alpha=0.2, linewidth=0.5)

legend_handles = [
    mpatches.Patch(color=C_GOOD,  label='Excellent retention  (< 8%)'),
    mpatches.Patch(color=C_SATIS, label='Acceptable  (8-12%)'),
    mpatches.Patch(color=C_POOR,  label='Concerning  (> 12%)'),
    mpatches.Patch(color=NAVY,    label='Actual turnover rate'),
    mlines.Line2D([], [], color='#888888', linewidth=2.5,
                  label='Company target  (white marker on bars)'),
]
ax.legend(handles=legend_handles, loc='lower right', fontsize=7.5, framealpha=0.9)
plt.tight_layout()
plt.show()
```

## Scenario C: Revenue Variance Bridge

A finance analyst is preparing a revenue bridge for the annual investor presentation, explaining how total revenue changed from the prior fiscal year to the current one. The bridge decomposes a net increase of $16M into five identified factors, some favorable and some unfavorable.

This scenario extends Example 20 by requiring the chart to represent both positive and negative variances simultaneously. Apply the decision framework before writing any code, and state explicitly how you will distinguish favorable from unfavorable contributions using color.

```python
# Scenario C Data
# A finance analyst is preparing a revenue bridge for the annual investor
# presentation, explaining how total revenue changed from the prior fiscal
# year to the current one. The bridge decomposes a net increase of $16M
# into five identified factors, some favorable and some unfavorable.
#
# This differs from Example 20 in two ways:
#   (1) The context is a period-over-period variance bridge, not a P&L bridge.
#   (2) The dataset contains BOTH positive (favorable) and negative (unfavorable)
#       variances, requiring a three-color encoding scheme.

bridge_labels = [
    'Prior Year\nRevenue', 'Volume\nEffect', 'Price\nEffect',
    'Mix\nEffect', 'New\nMarkets', 'Customer\nChurn', 'Current Year\nRevenue'
]

# Positive values = favorable variances; negative values = unfavorable variances.
# 'total' = starting or ending anchor value.
bridge_values    = [142, 24, -8, 6, 9, -15, 158]
bridge_bar_types = ['total', 'pos', 'neg', 'pos', 'pos', 'neg', 'total']
# Verify: 142 + 24 - 8 + 6 + 9 - 15 = 158
```

```python
# SCENARIO C -- Student Work Cell
#
# Analytical question: How did total revenue change from the prior fiscal year
# to the current year, and what were the direction and magnitude of each
# contributing factor?
#
# This scenario differs from Example 20 in that it contains BOTH favorable
# and unfavorable variances. The chart type and geometry are the same;
# the color encoding must now distinguish three categories:
#   -- starting and ending total bars
#   -- favorable variance bars (positive values)
#   -- unfavorable variance bars (negative values)
#
# Step 1: State the decision dimension (Time, Comparison, or Composition).
# Step 2: Identify the chart type from the session framework.
# Step 3: State how you will encode direction using color and which channel
#         carries magnitude. Note how the spacer-and-bar geometry differs
#         for a positive variance bar versus a negative one.
# Step 4: Produce the chart using the data provided above.
#
# Write your code below.
```

## Scenario C -- Instructor Solution

The cell below is the instructor solution. It should not be run or distributed to students prior to the class debrief.

```python
# ===========================================================
# SCENARIO C -- INSTRUCTOR SOLUTION
# ===========================================================
# Decision dimension: Composition (decomposition of change from a starting value).
# Chart type: Waterfall chart.
# Three-color scheme: navy = starting/ending totals,
#                     teal = favorable variances (positive),
#                     red-orange = unfavorable variances (negative).
# Connector line logic for mixed-direction variances:
#   After a positive or total bar: connector at bar top (= running total after)
#   After a negative bar:          connector at bar bottom (= running total after)

NAVY  = '#0A1E35'
TEAL  = '#1C7293'
RED_M = '#d6604d'

running  = 0
spacers  = []
heights  = []
bar_tops = []

for v, btype in zip(bridge_values, bridge_bar_types):
    if btype == 'total' and running == 0:
        spacers.append(0)
        heights.append(v)
        running = v
    elif btype == 'total':
        spacers.append(0)
        heights.append(running)
    elif v >= 0:
        spacers.append(running)
        heights.append(v)
        running += v
    else:
        new_running = running + v
        spacers.append(new_running)
        heights.append(abs(v))
        running = new_running
    bar_tops.append(spacers[-1] + heights[-1])

colors = []
for btype in bridge_bar_types:
    if btype == 'total': colors.append(NAVY)
    elif btype == 'pos': colors.append(TEAL)
    else:                colors.append(RED_M)

fig, ax = plt.subplots(figsize=(12, 5))
x     = np.arange(len(bridge_labels))
BAR_W = 0.55

ax.bar(x, spacers, width=BAR_W, color='none', alpha=0, zorder=1)
ax.bar(x, heights, bottom=spacers, width=BAR_W, color=colors, alpha=0.88, zorder=3)

# Connector: running total AFTER bar i
for i in range(len(bridge_labels) - 1):
    y_conn = spacers[i] if bridge_bar_types[i] == 'neg' else bar_tops[i]
    ax.plot(
        [x[i] + BAR_W / 2, x[i + 1] - BAR_W / 2],
        [y_conn, y_conn],
        color='#888888', linewidth=0.8, linestyle='--', zorder=2
    )

for i, (xi, sp, h, btype, v) in enumerate(
        zip(x, spacers, heights, bridge_bar_types, bridge_values)):
    if btype == 'total':
        lbl = f'${h}M'
    elif v >= 0:
        lbl = f'+${h}M'
    else:
        lbl = f'-${h}M'
    if h >= 12:
        ax.text(xi, sp + h / 2, lbl, ha='center', va='center',
                fontsize=8, color='white', fontweight='bold')
    else:
        ax.text(xi + 0.32, sp + h / 2, lbl, ha='left', va='center',
                fontsize=7.5, color=colors[i])

ax.set_xticks(x)
ax.set_xticklabels(bridge_labels, fontsize=8.5)
ax.set_ylabel('Revenue ($ Millions)', fontsize=9)
ax.set_ylim(0, 195)
ax.set_title(
    'Scenario C Solution: Revenue Variance Bridge, Prior Year to Current Year\n'
    'Navy = totals  |  Teal = favorable variances  |  Red-orange = unfavorable',
    fontsize=10
)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y', alpha=0.2, linewidth=0.5)
ax.axhline(y=0, color='#333333', linewidth=0.5)
plt.tight_layout()
plt.show()
```

## Scenario D: Sessions and Conversion Rate Over 90 Days

A marketing analyst is preparing a post-campaign report for the VP of Marketing. A digital campaign launched on Day 15. The analyst has 90 days of data: daily website sessions and daily conversion rate. Sessions are measured in count; conversion rate is measured as a proportion. The VP wants both metrics displayed on the same chart.

Note that the two measures are in different units. Apply the decision framework before writing any code.

```python
# Scenario D: Sessions and Conversion Rate Over 90 Days
np.random.seed(99)
days=np.arange(1,91)
pre_sess=3200+np.random.normal(0,150,14)
post_sess=4500+np.arange(76)*20+np.random.normal(0,200,76)
sessions_d=np.concatenate([pre_sess,post_sess]).astype(int)
pre_conv=0.032+np.random.normal(0,0.002,14)
post_conv=0.028+np.arange(76)*0.0003+np.random.normal(0,0.002,76)
conversion_d=np.clip(np.concatenate([pre_conv,post_conv]),0.01,0.08)
```

```python
# SCENARIO D -- Student Work Cell
#
# Analytical question: How did daily website sessions and conversion rate
# move over the 90-day period before and after the campaign launch?
#
# Note: sessions and conversion rate are in different units.
#
# Step 1: State the decision dimension (Time, Comparison, or Composition).
# Step 2: Identify the specific chart type from the session framework.
# Step 3: State the primary pre-attentive attribute for each variable and
#         explain why a single y-axis is not appropriate here.
# Step 4: Produce the chart using the data provided above.
#
# Write your code below.
```

## Scenario D -- Instructor Solution

The cell below is the instructor solution. It should not be run or distributed to students prior to the class debrief.

```python
# ===========================================================
# SCENARIO D -- INSTRUCTOR SOLUTION
# ===========================================================
# Decision dimension: Time (two measures tracked across a continuous time axis).
# Chart type: Dual-axis combination bar and line chart.
# Sessions (count) on the left axis as bars;
# conversion rate (proportion, displayed as %) on the right axis as a line.
# Two operationally related measures, different units, shared time axis.
# Campaign launch at Day 15 marked with a vertical reference line.

fig, ax1 = plt.subplots(figsize=(12, 5))

# Bars: daily session volume on left axis
ax1.bar(days, sessions_d, color='#aec7e8', alpha=0.75,
        label='Daily Sessions (left axis)', zorder=2)
ax1.set_ylabel('Daily Sessions', color='#2c5f8a', fontsize=9)
ax1.set_ylim(0, 6800)
ax1.tick_params(axis='y', labelcolor='#2c5f8a')
ax1.set_xlabel('Day', fontsize=9)

# Line: conversion rate on right axis (displayed as percentage)
ax2 = ax1.twinx()
ax2.plot(days, conversion_d * 100, color='#d6604d', linewidth=2.0,
         label='Conversion Rate % (right axis)', zorder=3)
ax2.set_ylabel('Conversion Rate (%)', color='#8b2020', fontsize=9)
ax2.set_ylim(0, 10)
ax2.tick_params(axis='y', labelcolor='#8b2020')

# Campaign launch marker
ax1.axvline(x=15, color='#1C7293', linewidth=1.5, linestyle='--',
            alpha=0.85, label='Campaign launch (Day 15)', zorder=4)

ax1.set_title(
    'Scenario D Solution: Daily Sessions and Conversion Rate, 90-Day Period\n'
    'Dual-axis: different units (count and proportion) on a shared time axis',
    fontsize=10
)

lines1, labs1 = ax1.get_legend_handles_labels()
lines2, labs2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labs1 + labs2, loc='upper left',
           fontsize=8, framealpha=0.9)
plt.tight_layout()
plt.show()
```
